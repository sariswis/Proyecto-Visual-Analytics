Lectura y muestra del archivo

```python
df = gpd.read_file('../data/original/municipios.geojson')
df.head()
```

**1.** Eliminar columnas innecesarias

```python
df = df.drop(columns=['VERSION', 'AREA', 'LATITUD', 'LONGITUD', 'TSP16_HOG', 'STP27_PERS'])
df.head()
```

**2.** Renombrar columnas

```python
df = df.rename(columns={
    'DPTO_CCDGO':'Código departamento',
    'MPIO_CCDGO':'Código municipio',
    'MPIO_CNMBR':'Municipio',
    'MPIO_CDPMP': 'DIVIPOLA',
})
df.head()
```

**3.** Mantener los municipios de Cundinamarca, Bogotá y Boyacá

```python
codigos_dpto = ["25", "11", "15"]  # Cundinamarca, Bogotá, Boyacá
df = df[df['Código departamento'].isin(codigos_dpto)]
df
```

**4.** Añadir la columna de departamento

```python
def get_department_name(code):
    department_names = {
        "11": "Bogotá",
        "15": "Boyacá",
        "25": "Cundinamarca"
    }
    return department_names.get(code, "Desconocido")

df['Departamento'] = df['Código departamento'].apply(get_department_name)
df = df[['Código departamento', 'Departamento', 'Código municipio', 'Municipio', 'DIVIPOLA', 'geometry']]
df.head()
```

Resultado final

```python
df
```

Exportar a CSV

```python
df.to_file("../data/preparada/municipios_final.geojson", driver="GeoJSON")
```
