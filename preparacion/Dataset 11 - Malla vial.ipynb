Lectura y muestra del archivo

```python
df = gpd.read_file('../data/original/vias.geojson')
df.head()
```

**1.** Eliminar columnas innecesarias

```python
df = df.drop(columns=['osm_id', 'code', 'ref', 'oneway', 'maxspeed', 'layer', 'bridge', 'tunnel'])
df.head()
```

**2.** Renombrar columnas

```python
df = df.rename(columns={
    'fclass':'Tipo',
    'name':'Nombre',
})
df.head()
```

**3.** Renombrar los tipos

```python
df['Tipo'] = df['Tipo'].replace({
    'trunk': 'Troncal',
    'primary': 'Primaria',
    'secondary': 'Secundaria',
})
```

Resultado final

```python
df
```

Exportar a CSV

```python
df.to_file("../data/preparada/vias_final.geojson", driver="GeoJSON")
```
