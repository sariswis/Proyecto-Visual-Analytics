# Cálculo de incidencia de las empresas en la calidad del aire

La probabilidad real que una empresa arroje una variable medida es:

$$Probabilidad_{Incidencia} = \frac{Probabilidad_{Combustible} * Ponderacion_{Combustible} + Probabilidad_{Fuente} * Ponderacion_{Fuente}}{9}$$

Donde: 

- $Probabilidad_{Combustible}$ y $Ponderacion_{Combustible}$ provienen del cruce de la variable y el tipo de combustible en la matriz del dataset 4
- $Probabilidad_{Fuente}$ y $Ponderacion_{Fuente}$ provienen del cruce de la variable y el tipo de fuente en la matriz del dataset 5

La incidencia de una empresa en una estación depende de la distancia empresa - estación ($distancia$) y la $Probabilidad_{Real}$

Para ello, se multiplica la probabilidad por un factor obtenido por la función gaussiana. Esta es una versión simplificada del Modelo Gaussiano de Dispersión de Contaminantes

$$Incidencia =  Probabilidad_{Real} * e^{-\frac{1}{2}  (\frac{distancia}{5})^2}$$
