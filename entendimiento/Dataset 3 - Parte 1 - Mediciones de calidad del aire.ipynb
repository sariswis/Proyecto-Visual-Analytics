```python
import pandas as pd
from ydata_profiling import ProfileReport
```

```python
from tabulate import tabulate

def print_table(df):
    print(tabulate(df, headers='keys', tablefmt='simple_outline'))
```

```python
filename = "../data/raw_data/bq-results-20251016-165053-1760633522998.csv"
df1 = pd.read_csv(filename)
```

```python
reporte = ProfileReport(df1)
reporte.to_file("archivos_generados/Reporte perfilamiento - Dataset 3 Original - 2024.html")
```

```python
non_numeric_attributes = df1.select_dtypes(exclude=['number']).columns.values

for col in non_numeric_attributes:
    attribute = df1[col].value_counts()
    df_attribute = pd.DataFrame({
        f"{col} values": attribute.index,
        'Count': attribute.values,
        'Percentage (%)': attribute.values * 100 / attribute.values.sum()
    }).sort_values(by='Count', ascending=False)

    print_table(df_attribute)
```
