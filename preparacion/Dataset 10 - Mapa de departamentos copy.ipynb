Lectura y muestra del archivo

```python
df = gpd.read_file('../data/original/departamentos.geojson')
df.head()
```

**1.** Eliminar columnas innecesarias

```python
df = df.drop(columns=['VERSION', 'AREA', 'LATITUD', 'LONGITUD', 'TSP16_HOG', 'STP27_PERS'])
df.head()
```

**2.** Renombrar columnas

```python
df = df.rename(columns={
    'DPTO_CCDGO':'Código departamento',
    'DPTO_CNMBR':'Departamento',
})
df.head()
```

**3.** Mantener los municipios de Cundinamarca, Bogotá y Boyacá

```python
codigos_dpto = ["25", "11", "15"]  # Cundinamarca, Bogotá, Boyacá
df = df[df['Código departamento'].isin(codigos_dpto)]
df
```

Resultado final

```python
df
```

Exportar a CSV

```python
df.to_file("../data/preparada/departamentos_final.geojson", driver="GeoJSON")
```
