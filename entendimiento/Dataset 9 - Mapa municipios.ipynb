# Entendimiento GeoJSONs

## Municipios

**Nombre de archivo:** municipios.geojson

Incluye los municipios colombianos

![Mpios](assets/Mpios.png)

Cada ítem tiene el código del departamento (DPTO_CCDGO), el código del municipio (MPIO_CCDGO), el nombre (MPIO_CNMBR), el código completo (MPIO_CDPMP), la versión, información geográfica y otros indicadores de TerriData
